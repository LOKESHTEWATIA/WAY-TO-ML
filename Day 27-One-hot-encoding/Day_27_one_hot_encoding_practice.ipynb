```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
```

```python
data = pd.read_csv('diamonds.csv')
data.head(5)
```

## 1. OneHotEncoding using Pandas

```python
pd.get_dummies(data, columns=['cut', 'color', 'clarity'],dtype=int)
```

```python
data_dropped = data.drop(columns=['carat', 'depth', 'table', 'x', 'y', 'z'])
display(data_dropped.head())
```

```python
data_encoded = pd.get_dummies(data_dropped, columns=['cut', 'color', 'clarity'], dtype=int)
display(data_encoded.head())
```

## 2. K-1 OneHotEncoding

```python
pd.get_dummies(data, columns=['cut', 'color', 'clarity'], drop_first=True,dtype=int)
```

## 3.OneHotEncoding with top categories

```python
counts = data['carat'].value_counts()
counts
```

```python
for i in data['carat']:
   print(i>3.00)
```

```python
thresold = 1.50
```

```python
filtered_data = data[data['carat'] < thresold]
display(filtered_data.head())
```

```python
carat_greater_than_3 = data[data['carat'] > 3.00]
display(carat_greater_than_3.head())
```

```python
filtered_data_cut = filtered_data['cut'].value_counts()
display(filtered_data_cut)
```

```python
average_price_by_cut = filtered_data.groupby('cut')['price'].mean().reset_index()
display(average_price_by_cut)
```
